```python
import google.generativeai as genai  # Google ki AI library import ki taaki Gemini use kar sakein
import os                            # System ke environment variables read karne ke liye os module import kiya
from dotenv import load_dotenv       # .env file se secret keys read karne ke liye load_dotenv import kiya

# 1. .env file se saari keys system environment me load karo
# override=True lagane se agar pehle se koi purani key memory me hai, toh wo nayi key se replace ho jayegi
load_dotenv(override=True)

# 2. .env file ke andar se GOOGLE_API_KEY ki value nikaalo aur Google AI library ko do taaki connection verify ho sake
genai.configure(api_key=os.getenv("GOOGLE_API_KEY"))

# 3. Gemini ka latest stable model ('gemini-2.5-flash') load karo jisse hume baat karni hai
model = genai.GenerativeModel('gemini-2.5-flash')

# Try-Except block taaki agar koi error aaye toh hamara program crash na ho
try:
    # 4. Model ko apna message bhejo aur response ka intezar karo
    response = model.generate_content("Say hello!")
    
    # 5. Agar sab sahi raha, toh AI ka reply screen par print kar do
    print(f"Response: {response.text}")
    
except Exception as e:
    # 6. Agar internet ka issue ho ya key galat ho, toh ye error message print karega
    print(f"Error: {e}")
```

```python
import os
from dotenv import load_dotenv

# Pehle make sure karein ki .env load ho chuka hai
load_dotenv(override=True)

# 1. Environment se sahi variable name ('GOOGLE_API_KEY') uthao
api_key = os.getenv("GOOGLE_API_KEY")

# 2. Validate karo ki key mili ya nahi
if api_key:
    # Key ke last 4 characters dikhao (security aur testing ke liye)
    print(f"✅ API Key loaded! ...{api_key[-4:]}")
else:
    # Agar variable nahi mila toh ye error aayega
    print("❌ ERROR: GOOGLE_API_KEY not found!")
    print("Check karein ki aapki .env file mein 'GOOGLE_API_KEY=your_key' sahi se likha hai.")
```

```python
# Gemini ko API key do
genai.configure(api_key=api_key)

print("🤖 Connecting to Gemini...")
print("-" * 50)

try:
    # Model select karo (FIXED: gemini-1.5-flash ko badal kar gemini-2.5-flash kiya)
    model = genai.GenerativeModel("gemini-2.5-flash")
    print("✅ Model loaded: gemini-2.5-flash")
    
    # Prompt bhejo
    prompt = "What is an AI agent? Explain in 2 sentences."
    print(f"\n📤 Prompt: {prompt}\n")
    
    # API call
    response = model.generate_content(prompt)
    
    # Response dikhao
    print("📥 Response:")
    print(response.text)
    print("\n" + "-" * 50)
    print("✅ SUCCESS! Gemini API working perfectly!")

except Exception as e:
    print(f"❌ Error: {e}")
    print("\nTroubleshooting Tips:")
    print("1. API key sahi hai? Google AI Studio se copy karo")
    print("2. Internet connected?")
    print("3. Free tier quota baki hai? Check karo: https://aistudio.google.com")
```

```python
# Different prompts try karo (Humne yahan comma fix kar diya hai)
prompts = [
    "What is an global warming ?",
    "What is the difference between prime numbers and whole numbers ?",
    "Explain temperature parameter in LLMs in simple words.",  # <-- FIXED: Yahan comma laga diya hai
    "Explain Nichola Tesla's universe frequency theory in simple words."
]

# Loop chalao taaki ek-ek karke saare prompts send ho sakein
# enumerate(..., 1) se hume 1, 2, 3, 4 index numbers milenge
for i, p in enumerate(prompts, 1):
    print(f"\n{'='*50}")
    print(f"📤 Prompt {i}: {p}")
    print(f"{'='*50}")
    
    try:
        # AI ko prompt bhejo
        response = model.generate_content(p)
        print(f"📥 Response {i}:")
        print(response.text)
    except Exception as e:
        # Agar kisi prompt par error aaye toh poora loop crash nahi hoga
        print(f"❌ Error in Prompt {i}: {e}")
        
    print()
```

```python
# Check installed packages
import subprocess  # Python se external terminal commands chalane ke liye import kiya
import sys         # Current running Python environment ka path nikaalne ke liye import kiya

print("📦 Installed Packages:")
print("-" * 30)

try:
    # 1. Command tayyar kiya: Current Python se 'pip list' nikaalo aur sirf 'generativeai' dhoondho
    command = f'"{sys.executable}" -m pip list | findstr generativeai'
    
    # 2. Command ko Windows terminal par execute kiya (shell=True se pipe '|' kaam karega)
    subprocess.check_call(command, shell=True)

except subprocess.CalledProcessError:
    # 3. Agar package list mein nahi mila, toh crash hone ke badle yeh message dikhao
    print("❌ 'generativeai' package not installed nahi mila!")
except Exception as e:
    # 4. Agar koi aur unexpected dikkat aayi, toh use yahan handle karo
    print(f"❌ different error: {e}")
```
